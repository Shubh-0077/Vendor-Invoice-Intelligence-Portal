

## Predicting Freight Cost

**Objective:** Predict freight cost for a vendor invoice using quantity and dollars, to improve cost forecasting, budgeting, and vendor negotiation.

* Freight is a non-trivial component of landed cost.
* Poor freight estimates distort margin and inventory planning.
* Automating freight estimation helps procurement teams forecast true cost before invoice arrival.

```python
# we have features like Quantity, Dollars, Freight for our task thats why we are using Vendor_df
vendor_df = pd.read_sql_query("SELECT * FROM vendor_invoice",conn)
vendor_df.head()
```

```python
vendor_df['freight_per_unit'] = vendor_df['Freight']/ vendor_df['Quantity']
vendor_df.head()
```

```python
x = vendor_df[['Quantity','Dollars']]
y = vendor_df['Freight']
```

```python
vendor_df.describe().round()
```

```python
from sklearn.model_selection import train_test_split 

from sklearn.linear_model import LinearRegression #(base model)
from sklearn.tree import DecisionTreeRegressor #(mid tier)
from sklearn.ensemble import RandomForestRegressor #(ensemble)

from sklearn.metrics import mean_absolute_error,mean_squared_error,r2_score
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.2,random_state=2)
```

```python
m1 = LinearRegression()
m1.fit(x_train,y_train)

m2 = DecisionTreeRegressor()
m2.fit(x_train,y_train) 

m3 = RandomForestRegressor()
m3.fit(x_train,y_train)
```

```python
def eval_model(model, x_test, y_test, model_name):
    y_pred = model.predict(x_test)

    mae = mean_absolute_error(y_test, y_pred)
    mse = mean_squared_error(y_test, y_pred)
    rmse = np.sqrt(mse)
    r2 = r2_score(y_test, y_pred)

    print(f"\n{model_name} Performance:")
    print(f"MAE : {mae:.2f}")
    print(f"MSE : {mse:.2f}")
    print(f"RMSE: {rmse:.2f}")
    print(f"R² Score: {r2:.4f}")
```

```python
eval_model(m1, x_test, y_test, "Linear Regression")
eval_model(m2, x_test, y_test, "Decision Tree Regressor")
eval_model(m3, x_test, y_test, "Random Forest Regressor")
```

### Model Evaluation Summary

- **Linear Regression** achieved the **highest R² score (0.9473)**, explaining approximately **94.73%** of the variance in freight cost.
- It also achieved the **lowest MAE (27.06)** and **lowest RMSE (174.37)**, indicating the smallest prediction errors among the evaluated models.
- Compared to the **Decision Tree Regressor** and **Random Forest Regressor**, **Linear Regression** demonstrated the best overall predictive performance.

### Conclusion

- The ideal regression model should have:
  - **Highest R² Score**
  - **Lowest MAE**
  - **Lowest RMSE**
- Based on these evaluation metrics, **Linear Regression** is selected as the **final model** for predicting freight costs.

```python
# lets visualise 
y_pred = m1.predict(x_test)

plt.figure(figsize=(6, 6))
plt.scatter(y_test, y_pred, alpha=0.7)

# Perfect prediction line
plt.plot(
    [y_test.min(), y_test.max()],
    [y_test.min(), y_test.max()],
    color="red",
    linestyle="--",
    linewidth=2
)

plt.xlabel("Actual Freight Cost")
plt.ylabel("Predicted Freight Cost")
plt.title("Actual vs Predicted Freight Cost")
plt.grid(True)
plt.show()
```
